```python
import numpy as np
import h5py
import matplotlib.pyplot as pl
import scipy
import os
import sys

from os.path import join
from util import get_index_from_rxList
```

```python
fname_in = 'test/summit_test_z_tx_50.0_buffer.h5'
with h5py.File(fname_in, 'r') as input_hdf:
    for key in input_hdf.keys():
        print(key)
    print('')
    print('attributes:')
    for key in input_hdf.attrs.keys():
        print(key, input_hdf.attrs[key])
    rxList = np.array(input_hdf['rxList'])
    rxPulses_z = np.array(input_hdf['rxPulses_z'])
    print('')
    nRx = rxList.shape[0]
    print(nRx)
    print(rxPulses_z.shape)
    nSamples = rxPulses_z.shape[1]
    dt = float(input_hdf.attrs['dt'])
    tspace = np.linspace(0, nSamples*dt, nSamples)
    
```

```python
fig = pl.figure(figsize=(8,8),dpi=120)
ax = fig.add_subplot(111)
ax.scatter(rxList[:,0], rxList[:,1])
ax.set_ylim(120, -20)
ax.grid()
pl.show()
```

```python
ii_rx = get_index_from_rxList(rxList, 10, 50)

fig = pl.figure(figsize=(8,5),dpi=120)
ax = fig.add_subplot(111)
ax.plot(tspace, rxPulses_z[ii_rx])
ax.grid()
pl.show()
```
